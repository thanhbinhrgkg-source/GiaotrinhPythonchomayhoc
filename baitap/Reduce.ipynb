```python
X = np.array([
    [3, 2, 7],
    [1, 4, 6]
], dtype=float)
X
```

```python
M = X.T @ X
eigv, eigvec = np.linalg.eigh(M)
orders = np.argsort(eigv)[::-1]
eigv[orders],
print(eigv[orders])
sigma = eigvec[orders][:,:2]
sigma
```

```python
reduce = X @ sigma
reduce
```
