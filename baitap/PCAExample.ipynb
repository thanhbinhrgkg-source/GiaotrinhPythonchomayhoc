```python
from sklearn.decomposition import PCA
import numpy as np
```

```python
M = np.array([
    [3, 7, 2],
    [6, 8, 1],
    [5, 4, 3],
    [9, 1, 7]
])
```

$X = M - mean(M)$


```python
X = M - M.mean(axis = 0)
X
```

$$ S = \frac{1}{n - 1} * M^T * M $$

```python
S = (1/(M.shape[0] - 1)) * X.T @ X
S
```

```python
eigenvalues, eigenvectors = np.linalg.eig(S)
indexes = np.argsort(eigenvalues)[::-1]
eigenvalues[indexes]
```

```python
k = 2
X @ eigenvectors[:,indexes][:,:k]
```

```python
pca = PCA(n_components = 2)
pca.fit(M)
pca.components_
```

```python
(M - M.mean(axis=0)) @ pca.components_.T
```

```python
pca.transform(M)
```
