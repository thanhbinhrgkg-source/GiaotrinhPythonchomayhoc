```python
from sklearn.cluster import DBSCAN
```

```python
X = [
    [2, 10],
    [2, 5],
    [8, 4],
    [5, 8],
    [7, 5],
    [6, 4],
    [1, 2],
    [4, 9],
    [3, 7.5],
]
dbscan = DBSCAN(eps=2, min_samples=3)
dbscan.fit(X)
dbscan.labels_
```
