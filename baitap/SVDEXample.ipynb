```python
from sklearn.decomposition import TruncatedSVD
import numpy as np
```

```python
M = np.array([
    [3, 7, 2],
    [6, 8, 1],
    [5, 4, 3],
    [9, 1, 7]
])
```

$$ S = M^T * M$$

```python
S = M.T @ M
S
```

Tính giá trị riêng và vector riêng

```python
eigenvalues, eigenvectors = np.linalg.eig(S)
eigenvectors
```

```python
indexes = np.argsort(eigenvalues)[::-1]
indexes
```

Chọn k vector riêng giảm dần theo giá trị riêng

```python
k = 2
V = eigenvectors[:, indexes][:,:k]
V
```
